```python
# verwerk hier jullie kalibratiemetingen

afstand = np.array([5, 10,15,20,25,30,35,40,45,50,55,60,65,70])  #in centimeter 
tijd = np.array([114,282,424,587,731,877,1022,1167,1312,1459,1612,1743,1878,2021]) #in micro seconde (heen en terug)
afstand_heenweer = afstand *2
foutmarge_tijd = 10 #in micro seconde

plt.figure()
plt.plot(afstand_heenweer,tijd,'k.', label = 'data')
plt.grid()
plt.xlabel("Afstand (cm)")
plt.ylabel("Tijd (us)")

def function (x,a,b):
    return  a*x +b
values, covariance = curve_fit(function, afstand_heenweer, tijd)
x_test = np.linspace(1, 150,100)
y_fit = function(x_test, *values)
plt.plot(x_test, y_fit, label = 'fit')
plt.legend()
plt.show()

a_fit = values[0]
b_fit = values[1]
print(b_fit)
print(a_fit)
```

Uitleg bij de algoritmes: 
Figuur 1. 
Ons originele idee was om simpelweg heel de x-as en y-as af te gaan met een verschil van 10 cn tussen de receiver en de sender. Hierna gingen we door naar de y-as en deden we hetzelfde. Dit zorgde ervoor dat we talloze ongeldige metingen hebben, wat niet handig is. Het vinden van het reflectiescherm moet sneller kunnen. 

Figuur 2: 
We beginnen op dezelfde manier, maar zodra er een hit is op de x-as ga je door naar de y-as op precies de helft van de R, wat ervoor zorgt dat je niet meer hoeft te 'zoeken' en meer geldige metingen kan krijgen.

Figuur 3: 
Voor nauwkeurigere metingen op de plek van het reflectiescherm, is het handig om ook andere afstanden tussen de receiver en de sender te doen dan alleen 10. 

Figuur 4: 
Als final version hebben we dit allemaal geintegreert. Hier hebben we dus ook extra metingen gedaan op de grootste range van de x-as, (groostet xs en kleinste xr) waarna je iterreert door xs en xr allebei steeds met 5 te verlagen per meting. Dit doe je ook op de y-as. Dit zorgt voor betere resulatten.
